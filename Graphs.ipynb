# Concepts

<details><summary><b>1. Graph Theory</b></summary>

**Definition**:  
The study of graphs, modeling relationships between objects.

**Mathematical notation**:
$$
G = (V, E)
$$
where:
- $V$ is a set of vertices,
- $E$ is a set of edges (pairs of vertices).

</details>

---

<details><summary><b>2. Graph / Network</b></summary>

**Definition**:  
Graphs with optional node/edge properties.

**Mathematical notation**:
$$
G = (V, E) \quad \text{with optionally} \quad w: E \to \mathbb{R}, \quad f: V \cup E \to \mathbb{R}^d
$$

</details>

---

<details><summary><b>3. Node / Vertex</b></summary>

**Definition**:  
A fundamental point in a graph.

**Mathematical notation**:
$$
v \in V
$$

</details>

---

<details><summary><b>4. Edge</b></summary>

**Definition**:  
A connection between two nodes.

**Mathematical notation**:
- Undirected: $e = \{u, v\}$
- Directed: $e = (u, v)$

</details>

---

<details><summary><b>5. Directed Graph</b></summary>

**Definition**:  
Graph with edges that have directions.

**Mathematical notation**:
$$
E \subseteq V \times V
$$

</details>

---

<details><summary><b>6. Undirected Graph</b></summary>

**Definition**:  
Graph with unordered edges.

**Mathematical notation**:
$$
E \subseteq \{ \{u, v\} \mid u, v \in V, u \neq v \}
$$

</details>

---

<details><summary><b>7. Adjacency</b></summary>

**Definition**:  
Two nodes are adjacent if directly connected.

**Mathematical notation**:
$$
\{u, v\} \in E \quad \text{or} \quad (u, v) \in E
$$

</details>

---

<details><summary><b>8. Degree</b></summary>

**Definition**:  
Number of edges incident to a node.

**Mathematical notation**:
- Undirected:
$$
\deg(v) = |\{ u \in V : \{u, v\} \in E\}|
$$
- Directed:
$$
\deg^-(v) = |\{ u : (u, v) \in E \}|, \quad \deg^+(v) = |\{ u : (v, u) \in E \}|
$$

</details>

---

<details><summary><b>9. Walk</b></summary>

**Definition**:  
Sequence of connected nodes and edges.

**Mathematical notation**:
$$
v_0, e_1, v_1, \ldots, e_k, v_k
$$

</details>

---

<details><summary><b>10. Trail</b></summary>

**Definition**:  
Walk with no repeated edges.

**Mathematical notation**:
$$
e_i \neq e_j \quad \text{for all} \quad i \neq j
$$

</details>

---

<details><summary><b>11. Path</b></summary>

**Definition**:  
Walk with no repeated vertices.

**Mathematical notation**:
$$
v_i \neq v_j \quad \text{for} \quad i \neq j
$$

</details>

---

<details><summary><b>12. Cycle</b></summary>

**Definition**:  
Path that starts and ends at the same node.

**Mathematical notation**:
$$
v_0 = v_k \quad \text{and} \quad v_i \neq v_j
$$

</details>

---

<details><summary><b>13. Circuit</b></summary>

**Definition**:  
Closed trail in directed graphs.

</details>

---

<details><summary><b>14. Bipartite Graph</b></summary>

**Definition**:  
Vertices can be divided into two disjoint sets.

**Mathematical notation**:
$$
V = U \cup W, \quad U \cap W = \emptyset
$$

</details>

---

<details><summary><b>15. Minimum Spanning Tree (MST)</b></summary>

**Definition**:  
Tree connecting all nodes with minimum weight.

**Mathematical notation**:
$$
\sum_{e \in E'} w(e) \quad \text{is minimized}
$$

</details>

---

<details><summary><b>16. Random Walk</b></summary>

**Definition**:  
Random transition between adjacent nodes.

**Mathematical notation**:
$$
\mathbb{P}(v_{i+1} = u \mid v_i = v) = \frac{1}{\deg(v)}
$$

</details>

---

<details><summary><b>17. Diffusion Sampling</b></summary>

**Definition**:  
Diffusion-based node sampling.

**Mathematical notation**:
$$
p_{t+1}(v) = \sum_{u : (u,v) \in E} \frac{p_t(u)}{\deg(u)}
$$

</details>

---

<details><summary><b>18. Graph Markov Models (GMMs)</b></summary>

**Definition**:  
Local dependency models.

**Mathematical notation**:
$$
\mathbb{P}(X_v \mid X_{V \setminus \{v\}}) = \mathbb{P}(X_v \mid X_{\text{Ne}(v)})
$$

</details>

---

<details><summary><b>19. Gaussian Graphical Models (GGMs)</b></summary>

**Definition**:  
Gaussian variables linked by conditional independence.

**Mathematical notation**:
$$
(\Sigma^{-1})_{uv} = 0
$$

</details>

---

<details><summary><b>20. Traversal</b></summary>

**Definition**:  
Systematic graph visiting.

</details>

---

<details><summary><b>21. Depth-First Traversal (DFS)</b></summary>

**Definition**:  
Explore deeply before backtracking.

</details>

---

<details><summary><b>22. Breadth-First Traversal (BFS)</b></summary>

**Definition**:  
Explore neighbors layer-by-layer.

</details>

---

<details><summary><b>23. Radial Visualization</b></summary>

**Definition**:  
Nodes placed on circles by distance.

</details>

---

<details><summary><b>24. Structural Equation Modeling (SEM)</b></summary>

**Definition**:  
Model complex variable relationships.

**Mathematical notation**:
$$
\mathbf{y} = \mathbf{B}\mathbf{y} + \mathbf{\Gamma} \mathbf{x} + \mathbf{\zeta}
$$

</details>

---

<details><summary><b>25. Networks</b></summary>

**Definition**:  
Graphs with attributes.

**Mathematical notation**:
$$
f: V \cup E \to \mathbb{R}^d
$$

</details>

---

<details><summary><b>26. Inductive Bias</b></summary>

**Definition**:  
Assumptions guiding learning models.

</details>

---

<details><summary><b>27. Aggregation (in GNNs)</b></summary>

**Definition**:  
Combining neighbor features.

**Mathematical notation**:
$$
h'(v) = \text{AGG}(\{ h(u) : u \in \text{Ne}(v) \})
$$

</details>

---

<details><summary><b>28. Graph Attention</b></summary>

**Definition**:  
Dynamic neighbor weighting.

**Mathematical notation**:
$$
h'(v) = \sum_{u \in \text{Ne}(v)} \alpha_{vu} h(u)
$$

</details>

---

<details><summary><b>29. Attributes</b></summary>

**Definition**:  
Features for nodes and edges.

**Mathematical notation**:
$$
f_V: V \to \mathbb{R}^d, \quad f_E: E \to \mathbb{R}^k
$$

</details>

---

<details><summary><b>30. Topology</b></summary>

**Definition**:  
The connectivity pattern of a graph, abstracted from geometry.
</details>

---

<details><summary><b>31. Phase Transition(Random Graphs)</b></summary>

**Definition**:  
Abrupt structural change in graph properties as a parameter crosses a threshold.

**Mathematical notation**:
In $ G(n, p) $, a giant component appears around:
$$
p = \frac{1}{n}
$$

</details>

---

<details><summary><b>32. Motif</b></summary>

**Definition**:  
Small, recurring subgraph patterns within larger networks.

</details>

---

<details><summary><b>33. Erdős–Rényi Model</b></summary>

**Definition**:  
A random graph model where each edge exists independently with probability $ p$.

**Mathematical notation**:
$$
G(n, p)
\quad \text{or} \quad G(n, m)
$$
depending on whether edge probability or number of edges is fixed.

</details>

# 📚 Graph Representations

A **graph representation** is the way we store and organize a graph $G = (V, E)$ inside a computer or in mathematical formulations.  
The right choice depends on the operations you want to perform: fast lookups, memory efficiency, graph algorithms, etc.

# Concepts

<details><summary><b>1. Adjacency Matrix</b></summary>

### Definition
A **square matrix** $A$ of size $|V| \times |V|$, where:

- For **undirected graphs**:
  $$
  A[i][j] =
  \begin{cases}
  1 & \text{if } \{i,j\} \in E \\\\
  0 & \text{otherwise}
  \end{cases}
  $$
- For **directed graphs**:
  $$
  A[i][j] =
  \begin{cases}
  1 & \text{if } (i,j) \in E \\\\
  0 & \text{otherwise}
  \end{cases}
  $$

If the graph is **weighted**, then $A[i][j] = w(i,j)$ where $w$ is the weight of edge $(i,j)$.

---

### Properties
| Property           | Notes                                        |
|--------------------|----------------------------------------------|
| Space Complexity   | $O(|V|^2)$ — always stores a full matrix     |
| Edge Lookup        | $O(1)$ — constant time to check an edge      |
| Best for           | Dense graphs (many edges, close to $|V|^2$)  |

---

### Example

Vertices = {0, 1, 2},  
Edges = { (0,1), (1,2) }

Adjacency matrix:
$$
A =
\begin{bmatrix}
0 & 1 & 0 \\\\
1 & 0 & 1 \\\\
0 & 1 & 0
\end{bmatrix}
$$

</details>

---

<details><summary><b>2. Adjacency List</b></summary>

### Definition
Each vertex stores a **list** (or set) of its **adjacent vertices**.

Formally, for each $v \in V$:
$$
\text{Adj}(v) = \{ u \in V : (v,u) \in E \}
$$

---

### Properties
| Property           | Notes                                        |
|--------------------|----------------------------------------------|
| Space Complexity   | $O(|V| + |E|)$ — efficient for sparse graphs |
| Edge Lookup        | $O(\text{degree}(v))$ — search through neighbors |
| Best for           | Sparse graphs (few edges)                   |

---

### Example

Vertices = ${0, 1, 2}$,  
Edges = ${ (0,1), (1,2) }$

Adjacency list:
$ 0: [1] 1: [0, 2] 2: [1] $


</details>

---

<details><summary><b>3. Edge List</b></summary>

### Definition
A **list of edges** where each edge is represented as a pair (or triple, if weighted).

Formally:
$$
\text{Edge List} = \{(u,v) \mid (u,v) \in E\}
$$
or
$$
\{(u,v,w) \mid (u,v) \in E, \text{ weight } w\}
$$

---

### Properties
| Property           | Notes                                    |
|--------------------|------------------------------------------|
| Space Complexity   | $O(|E|)$                                 |
| Edge Lookup        | $O(|E|)$ — must scan the list            |
| Best for           | Edge-based algorithms (e.g., Kruskal's MST) |

---

### Example

Vertices = $ {0, 1, 2}$ ,  
Edges = $ { (0,1), (1,2) } $

Edge list:

$[(0,1), (1,2)]$


If weighted:

$ [(0,1,5), (1,2,7)] $


</details>

---

<details><summary><b>4. Incidence Matrix</b></summary>

### Definition
A **matrix** $M$ of size $|V| \times |E|$, where:

- $M[i][j] = 1$ if vertex $i$ is incident to edge $j$,
- $M[i][j] = 0$ otherwise.

For **directed graphs**:
- $M[i][j] = 1$ if $i$ is the head (destination),
- $M[i][j] = -1$ if $i$ is the tail (source).

---

### Properties
| Property           | Notes                                       |
|--------------------|---------------------------------------------|
| Space Complexity   | $O(|V| \times |E|)$                         |
| Rare Usage         | Algebraic graph theory applications        |

---

### Example

Vertices = ${0, 1, 2}$,  
Edges = ${ (0,1), (1,2) }$

Incidence Matrix:
$$
M =
\begin{bmatrix}
1 & 0 \\\\
1 & 1 \\\\
0 & 1
\end{bmatrix}
$$

</details>

---

## Summary Table

| Representation     | Memory          | Edge Lookup       | Best for                 |
|---------------------|-----------------|-------------------|---------------------------|
| Adjacency Matrix     | $O(V^2)$         | $O(1)$            | Dense graphs              |
| Adjacency List       | $O(V + E)$       | $O(\deg(v))$       | Sparse graphs             |
| Edge List            | $O(E)$           | $O(E)$            | Edge-based algorithms     |
| Incidence Matrix     | $O(VE)$          | Rarely used       | Algebraic applications    |

---

## Computational Efficiency

- **Adjacency Matrix**: Very fast $O(1)$ edge lookup, but uses a lot of memory if sparse.
- **Adjacency List**: Efficient $O(\text{degree}(v))$ neighbor traversal.
- **Edge List**: Simple format; good for Kruskal’s MST.
- **Incidence Matrix**: Useful for linear algebra over graphs.

# **When to Choose Which Graph Representation?**

Choosing the right graph representation depends on:
- Graph **density** (sparse vs dense),
- **Operations** you care about (lookup, traversal, edge processing),
- **Memory** constraints,
- **Algorithmic needs** (some algorithms require specific formats).

| Situation                      | Recommended Representation   |
|---------------------------------|-------------------------------|
| Dense graph, fast lookup needed | Adjacency Matrix              |
| Sparse graph, neighbor traversal | Adjacency List               |
| Edge-centric algorithms         | Edge List                    |
| Algebraic/flow models           | Incidence Matrix             |

---

<details><summary><b>Adjacency Matrix</b></summary>

### Use an Adjacency Matrix when:
- You need **constant time** $O(1)$ edge existence checks.
- The graph is **dense** (close to $|V|^2$ edges).
- You want to use **matrix algebra** (e.g., eigenvalues, powers of adjacency matrix).

### Pros:
- Fast edge lookup $O(1)$.
- Good for mathematical computations (e.g., spectral graph theory).

### Cons:
- Memory heavy: $O(V^2)$ space even if few edges.

### Typical Use Cases:
- Dense social networks.
- Transition matrices in Markov chains.
- Graph-based machine learning with matrix operations.

</details>

---

<details><summary><b>Adjacency List</b></summary>

### Use an Adjacency List when:
- The graph is **sparse** ($|E| \ll |V|^2$).
- You mainly need **neighbor traversal** (e.g., DFS, BFS, GNN aggregation).

### Pros:
- Memory efficient $O(V + E)$.
- Fast to traverse all neighbors of a node.

### Cons:
- Edge existence checks take $O(\deg(v))$ time (may be slow if high-degree).

### Typical Use Cases:
- Web graphs (links between pages).
- Road maps (cities and roads).
- Sparse biological networks (e.g., protein-protein interactions).

</details>

---

<details><summary><b>Edge List</b></summary>

### Use an Edge List when:
- You mainly **process edges directly** (e.g., sorting edges by weight).
- You **don't care** about fast lookups.

### Pros:
- Very simple structure.
- Easy to load, sort, and process edges.

### Cons:
- Slow to check if an edge exists ($O(E)$ scan).

### Typical Use Cases:
- Kruskal's algorithm for Minimum Spanning Tree (MST).
- Graph construction from raw datasets.
- Streaming edge-based processing.

</details>

---

<details><summary><b>Incidence Matrix</b></summary>

### Use an Incidence Matrix when:
- You need **linear algebra formulations** (e.g., Graph Laplacian).
- You model **flow problems** or **network constraints**.

### Pros:
- Encodes relationships between vertices and edges explicitly.
- Useful for algebraic graph theory.

### Cons:
- Large memory footprint $O(V \times E)$.
- Rarely used for standard algorithms.

### Typical Use Cases:
- Electrical networks modeling (Kirchhoff's laws).
- Optimization in network flows.
- Algebraic graph theory research.

</details>

Let's build two biological networks —

* A Gene Expression Network (where edges link co-expressed genes),

* A Protein-Protein Interaction Network (PPI) (physical interactions between proteins),

— and then overlay them visually.

````python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Create Gene Expression Network (G_express)
G_express = nx.Graph()
G_express.add_edges_from([
    ('GeneA', 'GeneB'),
    ('GeneB', 'GeneC'),
    ('GeneC', 'GeneD'),
    ('GeneD', 'GeneE'),
    ('GeneA', 'GeneE')
])

# 2. Create Protein-Protein Interaction Network (G_ppi)
G_ppi = nx.Graph()
G_ppi.add_edges_from([
    ('GeneA', 'GeneC'),  # GeneA protein binds GeneC protein
    ('GeneB', 'GeneD'),
    ('GeneC', 'GeneE')
])

# 3. Overlay: Plot both together
pos = nx.spring_layout(G_express, seed=42)  # Same layout for both

plt.figure(figsize=(8,6))

# Draw Gene Expression edges
nx.draw_networkx_nodes(G_express, pos, node_color='lightblue', node_size=700, label='Genes')
nx.draw_networkx_edges(G_express, pos, edge_color='skyblue', width=2, label='Co-Expression')
nx.draw_networkx_labels(G_express, pos, font_size=10, font_weight='bold')

# Draw Protein-Protein edges on top
nx.draw_networkx_edges(G_ppi, pos, edge_color='red', style='dashed', width=2, label='Protein-Protein')

# Legend manually
plt.plot([],[], color='skyblue', label='Gene Expression (Co-Expression)')
plt.plot([],[], color='red', linestyle='--', label='Protein-Protein Interaction')

plt.legend(loc='best')
plt.title('Gene Expression Network and Protein-Protein Interaction Overlay')
plt.axis('off')
plt.show()
````

````python
# 4. Create a MultiGraph to merge both instead
G_merged = nx.MultiGraph()

# Add nodes
G_merged.add_nodes_from(G_express.nodes())

# Add edges from Gene Expression Network (coexpression edges)
for u, v in G_express.edges():
    G_merged.add_edge(u, v, interaction='coexpression')

# Add edges from Protein-Protein Interaction Network (ppi edges)
for u, v in G_ppi.edges():
    G_merged.add_edge(u, v, interaction='ppi')

# 5. Plot the MultiGraph

# Fixed layout so overlay looks good
pos = nx.spring_layout(G_merged, seed=42)

plt.figure(figsize=(8,6))

# Draw nodes
nx.draw_networkx_nodes(G_merged, pos, node_color='lightblue', node_size=700)

# Draw edges by type
# First coexpression edges
coexpr_edges = [(u, v, k) for u, v, k, d in G_merged.edges(keys=True, data=True) if d['interaction'] == 'coexpression']
nx.draw_networkx_edges(G_merged, pos, edgelist=coexpr_edges, edge_color='skyblue', width=2, label='Co-Expression')

# Then ppi edges
ppi_edges = [(u, v, k) for u, v, k, d in G_merged.edges(keys=True, data=True) if d['interaction'] == 'ppi']
nx.draw_networkx_edges(G_merged, pos, edgelist=ppi_edges, edge_color='red', style='dashed', width=2, label='PPI')

# Draw labels
nx.draw_networkx_labels(G_merged, pos, font_size=10, font_weight='bold')

# Manual legend
plt.plot([], [], color='skyblue', label='Gene Expression (Co-Expression)')
plt.plot([], [], color='red', linestyle='--', label='Protein-Protein Interaction (PPI)')

plt.legend(loc='best')
plt.title('Merged Gene Expression and Protein-Protein Interaction Network (MultiGraph)')
plt.axis('off')
plt.show()
````

# **Using a GNN Model to Infer PPI from Gene Expression**

We would like to train a GNN to predict whether two genes/proteins interact physically, using gene expression features (and possibly graph structure from co-expression).

**Input:**

  - A graph where nodes are genes.  
  - Node features = gene expression vectors (e.g., RNA-Seq expression values across samples).
  - Add known co-expression edges.

**Goal:**

Predict an edge label ("PPI" = 1, "No PPI" = 0) between gene pairs.

**Method:**

  - Use a GNN to embed nodes.
  - Predict edges based on learned embeddings.

````python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.data import Data
import numpy as np
import random

# we will leverage the initial example to demonstarte how GNNs work
# we will write custom functions for GNN and EdgePredictor
class GNN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels):
        super(GNN, self).__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, hidden_channels)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = F.relu(x)
        x = self.conv2(x, edge_index)
        return x

class EdgePredictor(torch.nn.Module):
    def __init__(self, embed_dim):
        super(EdgePredictor, self).__init__()
        self.fc1 = torch.nn.Linear(embed_dim*2, 64)
        self.fc2 = torch.nn.Linear(64, 1)

    def forward(self, u_emb, v_emb):
        z = torch.cat([u_emb, v_emb], dim=1)
        z = F.relu(self.fc1(z))
        z = torch.sigmoid(self.fc2(z))
        return z

# Start with the merged graph from the previous example- and label the edges for coexpression
for u, v in G_express.edges():
    G_merged.add_edge(u, v, interaction='coexpression')

# Also label the protein-protein interaction edges
for u, v in G_ppi.edges():
    G_merged.add_edge(u, v, interaction='ppi')

# Now Create node features:These are properties attributable to each node. In this case, these will be random "gene expression" vectors
gene_list = list(G_merged.nodes())
num_genes = len(gene_list)
num_features = 10  # 10 expression measurements(will be simulated at random)
node_features = np.random.randn(num_genes, num_features).astype(np.float32)

# Map gene names to indices
gene_to_idx = {gene: idx for idx, gene in enumerate(gene_list)}

# Next, Build PyG (torch_geometric) Data - this library requires a specific data structure (torch.tensor)
# Edge_index from coexpression edges only
coexpr_edges = [(gene_to_idx[u], gene_to_idx[v]) for u, v, d in G_merged.edges(data=True) if d['interaction'] == 'coexpression']
edge_index = torch.tensor(coexpr_edges, dtype=torch.long).t().contiguous()

# Node features
x = torch.tensor(node_features, dtype=torch.float)

# Specifiy positive and negative edge samples (for training purposes): hypothesis is if co-expression then ppi
positive_edges = [(gene_to_idx[u], gene_to_idx[v]) for u, v, d in G_merged.edges(data=True) if d['interaction'] == 'ppi']

# Negative samples: Random gene pairs not in known PPI
all_possible = [(i, j) for i in range(num_genes) for j in range(i+1, num_genes)]
existing_ppi = set((min(u,v), max(u,v)) for u,v in positive_edges)
negative_edges = []
for (i, j) in all_possible:
    if (i, j) not in existing_ppi and (i, j) not in coexpr_edges:
        negative_edges.append((i, j))

# Randomly pick available negatives
num_negatives = min(len(negative_edges), len(positive_edges))
negative_edges = random.sample(negative_edges,num_negatives)

# Combine positives and negatives
edge_pairs = positive_edges + negative_edges
labels = [1] * len(positive_edges) + [0] * len(negative_edges)

# Instantiate models
gnn = GNN(in_channels=num_features, hidden_channels=32)
predictor = EdgePredictor(embed_dim=32)

# Simple training loop
optimizer = torch.optim.Adam(list(gnn.parameters()) + list(predictor.parameters()), lr=0.01)
loss_fn = torch.nn.BCELoss()

for epoch in range(100):
    gnn.train()
    predictor.train()
    optimizer.zero_grad()

    node_embeddings = gnn(x, edge_index)

    preds = []
    for u, v in edge_pairs:
        h_u = node_embeddings[u].unsqueeze(0)
        h_v = node_embeddings[v].unsqueeze(0)
        pred = predictor(h_u, h_v)
        preds.append(pred)

    preds = torch.cat(preds, dim=0).squeeze()
    labels_tensor = torch.tensor(labels, dtype=torch.float)

    loss = loss_fn(preds, labels_tensor)
    loss.backward()
    optimizer.step()

    if (epoch+1) % 10 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")

# Inference Example: Predict PPI between GeneA and GeneC
geneA_idx = gene_to_idx['GeneA']
geneC_idx = gene_to_idx['GeneC']
h_geneA = node_embeddings[geneA_idx].unsqueeze(0)
h_geneC = node_embeddings[geneC_idx].unsqueeze(0)
ppi_prob = predictor(h_geneA, h_geneC).item()
print(f"\nPredicted PPI probability between GeneA and GeneC: {ppi_prob:.4f}")
````

# **Using the trained Model to create Synthetic Data**

The GNN model learned a latent space of genes based on expression + known PPIs, from this initial example we created.

We can now use the model to simulate new high-likelihood PPIs, on a new set of genes to grow our data of PPI and gene-coexpression networks from a tiny seed dataset!

We will need to:
1.	Initialize 50 genes with new random features (expression profiles)
2.	Embed new genes using trained GNN
3.	Predict edges between gene pairs
4.	Build a larger synthetic PPI network

````python
# Create 50 new genes
new_gene_list = [f"Gene{i}" for i in range(50)]

# Create new random gene expression features (similar to 5-gene feature structure)
new_num_genes = len(new_gene_list)
num_features = x.shape[1]  # use same number of features as before (10)

# Random new features
new_features = np.random.randn(new_num_genes, num_features).astype(np.float32)
new_x = torch.tensor(new_features, dtype=torch.float)
````

The GNN needs an edge_index to work, even if it's trivial (e.g., isolated nodes or small random coexpression edges).

We can specify "no prior edges" (nodes isolated), but the performance may drop.

So we will create a minimal random graph- this would be equivalent to random initialization of parameters prior to inference.

````python
# Create a random sparse graph structure (coexpression) for 50 genes
new_coexpr_edges = []
for i in range(new_num_genes):
    possible = [j for j in range(new_num_genes) if j != i]
    selected = random.sample(possible, 2)  # 2 neighbors per gene
    for j in selected:
        new_coexpr_edges.append((i, j))

# Convert to PyG format
new_edge_index = torch.tensor(new_coexpr_edges, dtype=torch.long).t().contiguous()
````

Now we can embed- the interaction strcuture between these 50 genes based on our trained GNN...

````python
# No need to retrain the GNN!
gnn.eval()

with torch.no_grad():
    new_node_embeddings = gnn(new_x, new_edge_index)

# Predict edges between all new gene pairs
predictor.eval()

new_possible_pairs = [(i, j) for i in range(new_num_genes) for j in range(i+1, new_num_genes)]

predicted_edges = []
predicted_scores = []

for (i, j) in new_possible_pairs:
    h_u = new_node_embeddings[i].unsqueeze(0)
    h_v = new_node_embeddings[j].unsqueeze(0)
    score = predictor(h_u, h_v).item()
    predicted_edges.append((i, j))
    predicted_scores.append(score)
````

FInally- we will build the synthetic PPI network and visualize it...

````python
threshold = 0.8  # can increase this for more confident predictions
synthetic_ppi_edges = [(i, j) for (i, j), score in zip(predicted_edges, predicted_scores) if score > threshold]

print(f"Generated {len(synthetic_ppi_edges)} synthetic PPIs among 50 genes!")
````

````python
G_synthetic = nx.Graph()

for idx, gene in enumerate(new_gene_list):
    G_synthetic.add_node(gene)

for (i, j) in synthetic_ppi_edges:
    G_synthetic.add_edge(new_gene_list[i], new_gene_list[j])

# Visualization
import matplotlib.pyplot as plt

pos = nx.spring_layout(G_synthetic, seed=42)
plt.figure(figsize=(12,10))
nx.draw(G_synthetic, pos, with_labels=True, node_color='lightgreen', node_size=500, font_size=8)
plt.title('Synthetic PPI Network Generated from GNN Trained on 5 Genes')
plt.show()
````
